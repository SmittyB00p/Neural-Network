```python
X = np.random.randn(5,3) # 5 instances with 3 features
Y = np.array([0,1,1,0,1], ndmin=2)
X, Y
```

```python
def initialize_parameters(X, layers=int, nodes=list):
    input_layer = X[0].reshape(1, X[0].size).T ## reshape to get nx1 vector

    parameters = {}

    ################### Initializes weights ###############
    for l, n in zip(range(layers), nodes):
        W = np.random.randn(n, input_layer.shape[0]) * 0.01
        b = np.ones((W.shape[0], 1))
        parameters[f"W{l+1}"] = W
        parameters[f"b{l+1}"] = b
        input_layer = b # output of layer l, which is the new input layer size

    return parameters
```

```python
params = initialize_parameters(X, 4,[4,3,2,1])
params
```

```python
def model(X, Y, params, layers):
    m = X.shape[0]
    cache = {} ## updates weights and bias terms throughout every example
    x = X[0].reshape(1, X[0].size).T ## reshape to get nx1 vector

    preds = [] ## hold predictions of each instance of X
    
    for i, instance in enumerate(X):
        # print(i, x, params[f"W{i+1}"], params[f"b{i+1}"])
        for l in range(layers):
            a = np.dot(params[f"W{l+1}"], x) + params[f"b{l+1}"]
            z = torch.relu(torch.tensor(a))
            dZ = np.where(z > 0, 1, 0)
            cache[f"a_{l+1}"] = a
            cache[f"z_{l+1}"] = z
            cache[f"dZ_{l+1}"] = dZ
            # print('------------------------------------')
            # print(a)
            # print('------------------------------------')
            x = a

        y_hat = torch.sigmoid(cache["z_4"].squeeze())
        
        preds.append(y_hat)

        try:
            x = X[i+1].reshape(1, X[i+1].size).T
        except:
            pass
        # print('------------------------------------')
        # print(x)
        # print('------------------------------------')

    preds = np.array(preds)

    # print(preds.shape, Y.shape)
    print(f"Predictions: {preds}")

    loss_fn = -np.sum((Y*np.log(preds)) + (1-Y*np.log(1-preds)))/m

    # #########################################
    print(f"Loss: {loss_fn}")


    ######### Back-Prop #####################
     
    

    #########################################
```
